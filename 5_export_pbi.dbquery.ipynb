```sql
CREATE OR REPLACE VIEW default.gold_demand_forecast_pbi AS
SELECT
    CAST(hour_ts AS TIMESTAMP) AS hour_ts,
    CAST(region AS STRING) AS region,
    CAST(actual_demand_mw AS DOUBLE) AS actual_demand_mw,
    CAST(predicted_demand_mw AS DOUBLE) AS predicted_demand_mw,
    CAST(created_at AS TIMESTAMP) AS created_at,
    CAST(year(hour_ts) AS INTEGER) AS year,
    CAST(month(hour_ts) AS INTEGER) AS month,
    CAST(dayofmonth(hour_ts) AS INTEGER) AS day,
    CAST(hour(hour_ts) AS INTEGER) AS hour
FROM delta.`/Volumes/workspace/default/gold/demand_forecast`;

SELECT * FROM default.gold_demand_forecast_pbi LIMIT 10;
```
